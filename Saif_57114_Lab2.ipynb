**BONUS TASK** Q) Create a GPA calculator based on what we have learned. (Using Lists, loops, if-else)

Course   | Credit Hours | Obtained Marks ( /100)

OOP      | 4            | 75
DB       | 3            | 80
Pak St.  | 2            | 90
English  | 2            | 99

```python
# FORMULA FOR GPA: GPA = (Σ (grade_point × credit_hours)) / (Σ credit_hours)

# GPA Calculator using Lists, Loop and If-Else

subjects = ["OOP", "DB", "Pak St.", "English"]
credits = [4, 3, 2, 2]

total_points = 0
total_credits = 0

for i in range(len(subjects)):
    marks = int(input("Enter marks for " + subjects[i] + ": "))

    if marks >= 85:
        gp = 4.0
    elif marks >= 75:
        gp = 3.5
    elif marks >= 65:
        gp = 3.0
    elif marks >= 55:
        gp = 2.5
    else:
        gp = 2.0

    total_points += gp * credits[i]
    total_credits += credits[i]

gpa = total_points / total_credits
print("Your GPA is:", round(gpa, 2))
```
