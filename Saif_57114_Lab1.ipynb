```python
def is_prime(n):
  if n==0 or n==1:
    return False
  for i in range (2,n):
    if n % i == 0:
      return False
  return True

count = 0

for i in range (1,20):
  if is_prime(i):
    print(count, "\t", i)
    count +=1
```
