**Steps of KNN**

1️⃣ Store training data

2️⃣ Calculate distance (Euclidean)

3️⃣ Pick K nearest points

4️⃣ Take majority vote

5️⃣ Assign class

```python
# Task 3: Calculate Accuracy

def accuracy(X_train, y_train, k):
    correct = 0

    for i in range(len(X_train)):
        test = X_train[i]

        X_temp = np.delete(X_train, i, axis=0)
        y_temp = np.delete(y_train, i)

        pred = knn(X_temp, y_temp, test, k)

        if pred == y_train[i]:
            correct += 1

    return correct / len(X_train)

print("Accuracy (k=3):", accuracy(X_train, y_train, 3))
```

```python
# EXAMPLE

import numpy as np
from collections import Counter

# training data
X_train = np.array([[1,2], [2,3], [3,3], [6,5], [7,7], [8,6]])
y_train = np.array(['A','A','A','B','B','B'])

# test point
test = np.array([5,5])

def distance(p1, p2):
    return np.sqrt(np.sum((p1 - p2)**2))

def knn(X_train, y_train, test, k):
    distances = []

    for i in range(len(X_train)):
        d = distance(test, X_train[i])
        distances.append((d, y_train[i]))

    distances.sort(key=lambda x: x[0])

    neighbors = distances[:k]

    labels = [label for _, label in neighbors]

    return Counter(labels).most_common(1)[0][0]

result = knn(X_train, y_train, test, 3)
print("Predicted Class:", result)
```
