```python
# Task 5: Bonus
# department average
dept_avg = df.groupby("Department")["Final_Exam_Score"].mean()

# check students above avg
result = []

for i in range(len(df)):
    dept = df.loc[i,"Department"]
    score = df.loc[i,"Final_Exam_Score"]

    if score > dept_avg[dept]:
        result.append((df.loc[i,"Name"], dept))

print(result)
```

```python
# Task 4: Sorting & Grouping
# sorting
df = df.sort_values(by=["GPA","Attendance"], ascending=[False, True])
print(df)

# grouping
group = df.groupby("Department").agg({
    "GPA":"mean",
    "Attendance":"mean",
    "StudentID":"count"
})

print(group)
```

```python
# Task 3: Feature Engineering
# Performance Level
def performance(gpa):
    if gpa >= 3.5:
        return "Excellent"
    elif gpa >= 3.0:
        return "Good"
    else:
        return "Needs Improvement"

df["Performance_Level"] = df["GPA"].apply(performance)

# Total Quiz
df["Total_Quiz"] = df["Quiz1"] + df["Quiz2"]

print(df)
```

```python
# Task 2: Data Cleaning
# remove duplicates
df = df.drop_duplicates()

# fill missing values
df = df.fillna(0)

# rename column
df = df.rename(columns={"FinalExam":"Final_Exam_Score"})

print(df)
```
