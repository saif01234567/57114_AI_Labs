```python
# **** LAB TASK 3 ****

#Q3) Create two sets (at least one set through constructor)
#set1 = {Even numbers (0~10)}
#set2 = {Positive integers (1~10)}

# set1 = Even numbers 0 to 10
set1 = {0,2,4,6,8,10}

# set2 using constructor
set2 = set(range(1,11))

print("Set1:", set1)
print("Set2:", set2)

# a) Remove second and fifth element of set2
temp = list(set2)

# remove second and fifth (index 1 and 4)
temp.pop(4)
temp.pop(1)

set2 = set(temp)
print("Updated set2:", set2)

# b) set3 = {intersection of set2 and set1}
set3 = set1.intersection(set2)
print("Intersection:", set3)
```

```python
#  **** LAB TASK 1 ****

#Q1) Create Numpy array, split into 3 arrays then calculate sum of all arrays numbers.

import numpy as np

# Create array
arr = np.arange(1, 10)   # numbers 1 to 9

# Split into 3 arrays
split_arrays = np.array_split(arr, 3)

print("Split Arrays:", split_arrays)

# Calculate total sum
total_sum = 0
for a in split_arrays:
    total_sum += np.sum(a)

print("Total Sum:", total_sum)
```
